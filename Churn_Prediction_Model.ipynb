```python
import pandas as pd
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.metrics import recall_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('churn_data.csv')
```

```python
df.head()
```

```python
df = df.drop('Unnamed: 0',axis=1)
```

```python
df.head()
```

```python
#Creating X and Y
X = df.drop('Churn',axis=1)
y = df['Churn']
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.2
)
```

### Decision Tree Classifier

```python
dt_model = DecisionTreeClassifier(criterion ='gini',random_state= 100,max_depth=6,min_samples_leaf= 8)
```

```python
dt_model.fit(X_train,y_train)
```

```python
y_pred = dt_model.predict(X_test)
```

```python
y_pred
```

```python
dt_model.score(X_test,y_pred)
```

```python
print(classification_report(y_test,y_pred,labels = [0,1]))
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
#To resolve the issue of IMBALANCED DATA
from imblearn.combine import SMOTEENN

smote_enn = SMOTEENN(random_state=42)
X_resampled, y_resampled = smote_enn.fit_resample(X, y)
```

```python
Xr_train,Xr_test,yr_train,yr_test = train_test_split(
    X_resampled,y_resampled,test_size=0.2
)
```

```python
model_dt_smote = DecisionTreeClassifier(criterion = 'gini',random_state = 100,max_depth=6,min_samples_leaf= 8)
```

```python
model_dt_smote.fit(Xr_train,yr_train)
```

```python
y_pred_smote = model_dt_smote.predict(Xr_test)
```

```python
print(classification_report(yr_test,y_pred_smote,labels = [0,1]))
```
