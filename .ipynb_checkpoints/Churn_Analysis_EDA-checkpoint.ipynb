```python
new_df = new_df.dropna(subset=['TotalCharges'])
```

```python
#Max Tenure
print(new_df['tenure'].max())
```

```python
#Group the tenure in bins of 12months
labels = ["{0} - {1}".format(i,i+11) for i in range (1,72,12)]

new_df['tenure_group'] = pd.cut(new_df.tenure,range(1,79,12),right=False, labels = labels)
```

```python
new_df['tenure_group'].value_counts()
```

```python
pd.crosstab(new_df['gender'], new_df['Churn'])
```

```python
new_df.drop(columns = ['customerID','tenure'],axis = 1,inplace = True)
new_df.head()
```

#EDA-->Exploratory Data Analysis

```python
for i, predictor in enumerate(
    new_df.drop(columns=['Churn',
                         'TotalCharges',
                         'MonthlyCharges'])
):

    plt.figure(i)

    sns.countplot(
        data=new_df,
        x=predictor,
        hue='Churn'
    )

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()
```

```python
#Encoding Target Variable,Yes = 1,No=0

new_df['Churn'] = new_df['Churn'].map({'Yes':1,'No':0})
```

```python
new_df['Churn']
```

```python
#Encoding  --> Categorical vals to numerical vals
new_df_dummies = pd.get_dummies(new_df,dtype = int,drop_first = True)
new_df_dummies.head()
```

```python
sns.lmplot(
    data=new_df,
    x='MonthlyCharges',
    y='TotalCharges'
)
```

```python
sns.kdeplot(
    data=new_df_dummies,
    x='MonthlyCharges',
    hue='Churn',
    fill=True
)

plt.show()
```

```python
#Churn is high when monthly charges are high
```

```python
plt.figure(figsize=(20,8))
new_df_dummies.corr()['Churn'].sort_values(ascending= False).plot(kind='bar')
```

*DERIVED INSIGHT:*

**HIGH CHURN** seen in case of **Month to Month Contracts**.**No online security,No Tech Support,First Year of subscription and Fiber Optics Internet**


**LOW** churn seen in case of **Long Term contracts,Subscription without Internet Service and the customers engaged for 5+ years**

Factors like **Gender,Availability of Phone Services and # of multiple lines have almost NO impact on Churn**

Also evident from **Heatmap** Below

```python
plt.figure(figsize=(12,12))
sns.heatmap(new_df_dummies.corr(),cmap="Paired")
```

#BIVARIATE ANALYSIS

```python
new_df1_target0 = new_df.loc[new_df["Churn"] == 0]
new_df1_target1 = new_df.loc[new_df["Churn"] == 1]
```

```python
def uniplot(df, col, title, hue=None):

    sns.set_style('whitegrid')

    sns.set_context('talk')

    plt.rcParams["axes.labelsize"] = 18

    plt.rcParams["axes.titlesize"] = 22

    plt.rcParams["axes.titlepad"] = 22


    fig, ax = plt.subplots(figsize=(14,8))

    sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index,
        hue=hue,
        palette='bright'
    )

    plt.xticks(rotation=45)

    plt.title(title)

    plt.tight_layout()

    plt.show()
```

```python
uniplot(new_df1_target1,col="Partner", title = 'Distribution of Gender for Churned Customers',hue='gender')
```

```python
uniplot(new_df1_target0,col="Partner", title = 'Distribution of Gender for Non Churned Customers',hue='gender')
```

```python
uniplot(new_df1_target1,col="TechSupport", title = 'Distribution of TechSupport for Churned Customers',hue='gender')
```

```python
uniplot(
    new_df1_target1,
    col="SeniorCitizen",
    title='Distribution of SeniorCitizen for Churned Customers',
    hue='gender'
)
```

```python
uniplot(
    new_df1_target1,
    col="PaymentMethod",
    title='Distribution of Payment Method for Churned Customers',
    hue='Churn'
)
```

**CONCLUSION**

1.Electronic check medium are the highest churners

2.No Online Security,No Tech Support are high churners

3.Non Senior Citizens are high churners
